```python
X = df.iloc[:, :7]
y = df.iloc[:, -1]
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
print("y_train counts:\n", y_train.value_counts())
print("\ny_train proportions:\n", y_train.value_counts(normalize=True))

print("\ny_test counts:\n", y_test.value_counts())
print("\ny_test proportions:\n", y_test.value_counts(normalize=True))
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Fit on training data and transform it
X_train_scaled = scaler.fit_transform(X_train)

# Only transform test data (using training mean/std to prevent data leakage)
X_test_scaled = scaler.transform(X_test)
```

```python
import numpy as np
print("X_train_scaled shape:", X_train_scaled.shape)
print("X_test_scaled shape:", X_test_scaled.shape)
print("Mean of scaled training features:", np.mean(X_train_scaled, axis=0))
```

## Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(
    random_state=42,
    class_weight='balanced'
)
```

```python
model.fit(X_train, y_train)
```

```python
model.fit(X_train_scaled, y_train)
```

```python
y_pred = model.predict(X_test_scaled)
print(y_pred)
```

```python
from sklearn.metrics import confusion_matrix
```

```python
cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
from sklearn.metrics import classification_report
```

```python
report = classification_report(y_test, y_pred)

print(report)
```

```python
from sklearn.metrics import roc_auc_score
```

```python
y_prob = model.predict_proba(X_test_scaled)[:, 1]
```

```python
roc_auc = roc_auc_score(y_test, y_prob)

print("ROC-AUC Score:", roc_auc)
```
